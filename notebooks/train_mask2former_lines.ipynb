# Mask2Former line training

This notebook starts training from scratch when no local checkpoint exists. If a checkpoint is found in `output_dir`, it resumes from that local checkpoint. Fresh training uses the HuggingFace entry only as an architecture/image-processor config source; model weights are randomly initialized with `from_config`.

```python
from pathlib import Path
import sys

cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd.parent if cwd.name == "notebooks" else cwd
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

PROJECT_ROOT
```

```python
from utils import TrainConfig, find_resume_checkpoint, run_training
```

```python
OUTPUT_DIR = PROJECT_ROOT / "mask2former_lines_scratch"
resume_checkpoint = find_resume_checkpoint(OUTPUT_DIR)
print("Resume checkpoint:", resume_checkpoint if resume_checkpoint else "none, start from scratch")

config = TrainConfig(
    model_config="facebook/mask2former-swin-tiny-coco-instance",
    train_img_dir=str(PROJECT_ROOT / "handwritten_essay_v2_east_lines/train_img"),
    train_gt_dir=str(PROJECT_ROOT / "handwritten_essay_v2_east_lines/train_gt"),
    val_img_dir=str(PROJECT_ROOT / "handwritten_essay_v2_east_lines/test_img"),
    val_gt_dir=str(PROJECT_ROOT / "handwritten_essay_v2_east_lines/test_gt"),
    output_dir=str(OUTPUT_DIR),
    resume_from_checkpoint=str(resume_checkpoint) if resume_checkpoint else None,
    auto_resume=True,
    image_size=512,
    load_max_side=1024,
    epochs=50,
    batch_size=1,
    accumulation_steps=8,
    max_instances_per_image=32,
    lr=1e-5,
    weight_decay=1e-4,
    num_workers=0,
    checkpoint_interval_steps=100,
    preview_count=3,
    metric_images=20,
)

config
```

```python
result = run_training(config)
result
```
